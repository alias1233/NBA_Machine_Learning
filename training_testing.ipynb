```python
import torch
from matplotlib import pyplot as plt

from src.training.loss_functions.gaussian_smoothed_cross_entropy_loss import GaussianSmoothedCrossEntropyLoss
```

```python
torch.set_printoptions(sci_mode=False)
```

```python
gaussian_loss = GaussianSmoothedCrossEntropyLoss(60, 1)
```

```python
distribution = gaussian_loss._make_soft_labels(torch.Tensor([10]))[0]
```

```python
plt.figure(figsize=(10, 4))
plt.bar(
    range(len(distribution)),
    distribution,
    color='black'
    )

plt.xlabel("Value")
plt.ylabel("Probability")
plt.title(f"Distribution")
plt.show()
```
