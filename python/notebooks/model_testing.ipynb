```python
import torch
from matplotlib import pyplot as plt

from nba_ml.machine_learning.losses.gaussian_smoothed_cross_entropy import GaussianSmoothedCrossEntropyLoss
```

```python
torch.set_printoptions(sci_mode=False)
```

```python
gaussian_loss = GaussianSmoothedCrossEntropyLoss(60, 0.5)
```

```python
distribution = gaussian_loss._make_soft_labels(torch.Tensor([10]))[0]
```

```python
plt.figure(figsize=(10, 4))
plt.bar(
    range(len(distribution)),
    distribution,
    color='black'
    )

plt.xlabel("Value")
plt.ylabel("Probability")
plt.title(f"Distribution")
plt.show()
```

```python
from datetime import date
from matplotlib import pyplot as plt

from src.nba_ml import season
from src.nba_ml.entities import Game
from src.nba_ml.models.player_stats import inference, specifications
from src.nba_ml.stores import players, teams
from src.nba_ml.odds import calculate_odds
```

```python
removed_core_teammates: set[int] = set([
    players.find_by_name('Austin Reaves')[0].id
])

added_core_teammates: set[int] = set([
    # players.find_by_name('Luka Doncic')[0].id
])

predicted_probabilities = inference.predict(
    specifications.POINTS_MODEL,
    players.find_by_name('Luka Doncic')[0].id,
    Game(
        date(2026, 6, 30),
        teams.find('spurs').id,
        teams.find('lakers').id,
        season.SeasonType.PLAYOFFS
    ),
    removed_core_teammates,
    added_core_teammates
)
```

```python
predicted_probabilities
```

```python
plt.figure(figsize=(10, 4))
plt.bar(
    range(len(predicted_probabilities)),
    predicted_probabilities,
    color='black'
    )

plt.xlabel("Value")
plt.ylabel("Probability")
plt.title(f"Predictions")
plt.show()
```
